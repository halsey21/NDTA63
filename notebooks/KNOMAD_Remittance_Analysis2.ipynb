```python
# Get the South Africa rows
sa_inward = south_africa_inward.iloc[0]
sa_outward = south_africa_outward.iloc[0]

# Years required for the analysis
years = [str(year) for year in range(2000, 2025)]

# Function to safely find each year in the row
def get_year_value(row, year):
    for column in row.index:
        if str(column).strip() == str(year):
            return row[column]
    return None

# Create a clean dataframe
remittance_df = pd.DataFrame({
    "Year": [int(year) for year in years],
    "Inward": [get_year_value(sa_inward, year) for year in years],
    "Outward": [get_year_value(sa_outward, year) for year in years]
})

# Convert values to numbers
remittance_df["Inward"] = pd.to_numeric(remittance_df["Inward"], errors="coerce")
remittance_df["Outward"] = pd.to_numeric(remittance_df["Outward"], errors="coerce")

print("Remittance dataframe created successfully.")

display(remittance_df)
```

<h1 style="color:#17365D;"> 4. Database Integration </h1>

<h2 style="color:#2F75B5;">  4.1 Build and Query a Database </h2>

The KNOMAD South Africa inward and outward remittance data is stored in an SQLite database. The database is created to allow the data to be stored, queried, updated and deleted safely.

```python
import sqlite3

connection = sqlite3.connect("knomad_remittance.db")

cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS remittances (
    Year INTEGER PRIMARY KEY,
    Inward REAL,
    Outward REAL
)
""")

for _, row in remittance_df.iterrows():
    cursor.execute("""
    INSERT OR REPLACE INTO remittances (Year, Inward, Outward)
    VALUES (?, ?, ?)
    """, (
        int(row["Year"]),
        float(row["Inward"]),
        float(row["Outward"])
    ))

connection.commit()

print("Database and remittance table created successfully.")
```

<h2 style="color:#2F75B5;"> 4.2 Query the Database </h2>

A SQL query is used to retrieve the South African inward and outward remittance records from the database.

```python
cursor.execute("""
SELECT Year, Inward, Outward
FROM remittances
ORDER BY Year
""")

records = cursor.fetchall()

for record in records:
    print(record)
```

<h2 style="color:#2F75B5;"> 4.3 Update and Delete Records Safely</h2>

SQL UPDATE and DELETE statements are used with specific conditions to ensure that only the intended records are affected. The changes are committed to the database after the operations are completed.

```python
cursor.execute("""
SELECT Year, Inward, Outward
FROM remittances
WHERE Year = 2024
""")

print("Before update:")
print(cursor.fetchone())
```

```python
cursor.execute("""
SELECT *
FROM remittances
WHERE Year = 2024
""")

print(cursor.fetchone())
```

```python
cursor.execute("""
UPDATE remittances
SET Outward = ?
WHERE Year = ?
""", (1000.00, 2024))

cursor.execute("""
SELECT Year, Inward, Outward
FROM remittances
WHERE Year = 2024
""")

print("After update:")
print(cursor.fetchone())
```

```python
cursor.execute("""
UPDATE remittances
SET Outward = ?
WHERE Year = ?
""", (993.22778264, 2024))

connection.commit()

print("Updated record restored successfully.")
```

<h2 style="color:#2F75B5;"> 4.4 Delete a Database Record </h2>

An SQL DELETE operation is demonstrated using a database transaction. The deletion is tested and then rolled back so that the original dataset remains unchanged.

```python
connection.execute("BEGIN")

cursor.execute("""
SELECT Year, Inward, Outward
FROM remittances
WHERE Year = 2000
""")

print("Record before deletion:")
print(cursor.fetchone())

cursor.execute("""
DELETE FROM remittances
WHERE Year = 2000
""")

cursor.execute("""
SELECT Year, Inward, Outward
FROM remittances
WHERE Year = 2000
""")

print("Record after deletion:")
print(cursor.fetchone())

connection.rollback()

print("Deletion rolled back. Original database restored.")
```

Conditional formatting is used to highlight higher and lower remittance values. This makes important patterns in the data easier to identify.

```python
analysis_df.style.background_gradient(
    subset=["Inward", "Outward", "Difference"]
)
```

<h2 style="color:#2F75B5;"> 5.3 Create Charts and Summarise Findings </h2>

Charts are used to present the remittance data visually and make comparisons between inward and outward remittance flows easier.

```python
plt.figure(figsize=(12, 6))

plt.bar(
    analysis_df["Year"] - 0.2,
    analysis_df["Inward"],
    width=0.4,
    label="Inward Remittance"
)

plt.bar(
    analysis_df["Year"] + 0.2,
    analysis_df["Outward"],
    width=0.4,
    label="Outward Remittance"
)

plt.title("South Africa Inward and Outward Remittances")
plt.xlabel("Year")
plt.ylabel("Remittance (US$ million)")
plt.legend()
plt.xticks(analysis_df["Year"], rotation=45)

plt.tight_layout()
plt.show()
```

### Summary of Findings

The analysis shows that outward remittances were generally higher than inward remittances throughout the period from 2000 to 2024. Both types of remittances increased substantially during the earlier years and reached their highest levels around 2011. After this period, the values fluctuated, with a noticeable decline around 2020 followed by a recovery in 2021.

The comparison indicates that South Africa consistently had higher outward than inward remittance flows during the period analysed.
